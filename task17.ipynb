```python
import pandas as pd
```

```python
df = pd.read_csv ("IRIS.csv")
```

```python
# Display the first 5 rows to understand the structure
print("--- First 5 Rows of the Dataset ---")
print(df.head(), "\n")
```

```python
# Selecting specific rows (index 0 to 4) and specific columns by name
print("Using loc: Rows 0 to 4 and specific columns ")
subset_loc = df.loc[0:4, ['sepal_length', 'sepal_width', 'species']]
print(subset_loc, "\n")
```

```python
# Select all rows for a single column using label
print(" Using loc: Single column ('species') ")
single_col_loc = df.loc[:, 'species']
print(single_col_loc.head(), "\n")
```

```python
#iloc
```

```python
# Select rows from index 0 to 4 and columns from position 0 to 2 (exclusive of index 3)
print("--- Using iloc: Rows 0:5 and columns 0:3 ---")
subset_iloc = df.iloc[0:5, 0:3]
print(subset_iloc, "\n")
```

```python
# Select specific non-contiguous rows and columns using lists of integers
print("--- Using iloc: Specific rows [0, 2, 4] and columns [0, 3] ---")
custom_iloc = df.iloc[[0, 2, 4], [0, 3]]
print(custom_iloc, "\n")
```
